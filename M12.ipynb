```python
#nomor 1
class Karyawan:
  def __init__(self, nama, gaji):
    self.nama = nama
    self.gaji = gaji
  def tampilkan_gaji(self):
    print(f"Karyawan {self.nama} memiliki gaji {self.gaji}.")

karyawan1 = Karyawan("Siti", 5000000)
karyawan2 = Karyawan("Budi", 6000000)
karyawan3 = Karyawan("Cici", 7000000)
karyawan4 = Karyawan("Nina", 8000000)
karyawan5 = Karyawan("Tata", 9000000)

karyawan1.tampilkan_gaji()
karyawan2.tampilkan_gaji()
karyawan3.tampilkan_gaji()
karyawan4.tampilkan_gaji()
karyawan5.tampilkan_gaji()
```

```python
#NOMOR 2
class Karyawan:
  def _init_(self, nama, gaji):
    self.nama = nama
    self.gaji = gaji
    self.energi = 100

  def lembur(self):
    print(f"{self.nama} sedang lembur!")
    self.gaji += 200000
    self.energi -= 30
    print (f"Sisa energi: {self.energi} dan sisa gaji: {self.gaji}" )

karyawan1 = Karyawan("Siti", 5000000)
karyawan2 = Karyawan("Budi", 6000000)
karyawan3 = Karyawan("Cici", 7000000)
karyawan4 = Karyawan("Nina", 8000000)
karyawan5 = Karyawan("Tata", 9000000)

karyawan1.lembur()
karyawan2.lembur()
karyawan3.lembur()
karyawan4.lembur()
karyawan5.lembur()
```

```python
#NOMOR 4
class RekeningBank:
    def _init_(self, nama_pemilik, saldo=0):
        self.nama_pemilik = nama_pemilik
        self.saldo = saldo

    def menabung(self, nominal):
        self.saldo += nominal
        print(f"Berhasil menabung Rp{nominal}. Saldo sekarang: Rp{self.saldo}")

    def tarik_tunai(self, nominal):
        if self.saldo >= nominal:
            self.saldo -= nominal
            print(f"Berhasil menarik Rp{nominal}. Saldo sekarang: Rp{self.saldo}")
        else:
            print("Maaf, saldo tidak mencukupi.")


rekening = RekeningBank("Nama Kamu", 150000)
rekening.tarik_tunai(50000)
rekening.tarik_tunai(200000)
rekening.menabung(300000)
```
